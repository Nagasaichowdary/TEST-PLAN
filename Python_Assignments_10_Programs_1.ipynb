## 1. Print all prime numbers between input range

```python
first = int(input("Enter starting value: "))
last = int(input("Enter ending value: "))

primes = []

for value in range(max(2, first), last + 1):
    is_prime = True
    for divisor in range(2, value):
        if value % divisor == 0:
            is_prime = False
            break
    if is_prime:
        primes.append(value)

print("Prime numbers:", *primes)
```

## 2. Factorial using recursion

```python
def find_factorial(value):
    if value < 0:
        return None
    if value <= 1:
        return 1
    return value * find_factorial(value - 1)

number = int(input("Enter a number: "))
answer = find_factorial(number)

if answer is None:
    print("Factorial is not defined for negative numbers.")
else:
    print("Factorial:", answer)
```

## 3. Square of numbers using lambda

```python
values = [int(item) for item in input("Enter numbers separated by space: ").split()]

make_square = lambda value: value ** 2
squared_values = [make_square(value) for value in values]

print("Original numbers:", values)
print("Squares:", squared_values)
```

## 4. Find the second largest element in a list

```python
values = [int(item) for item in input("Enter numbers separated by space: ").split()]

largest = max(values)
remaining = [value for value in values if value != largest]

if remaining:
    print("Second largest element:", max(remaining))
else:
    print("There is no second largest element.")
```

## 5. Count frequency of characters in a string

```python
text = input("Enter a string: ")
counts = {}

for symbol in text:
    if symbol in counts:
        counts[symbol] += 1
    else:
        counts[symbol] = 1

print("Character frequency:")
for symbol in counts:
    print(symbol, ":", counts[symbol])
```

## 6. Calculate area of a circle using math library

```python
from math import pi

radius = float(input("Enter radius of the circle: "))
circle_area = pi * (radius ** 2)

print("Area of the circle:", circle_area)
```

## 7. Reverse a string without using built-in reverse

```python
text = input("Enter a string: ")

characters = []
for position in range(len(text) - 1, -1, -1):
    characters.append(text[position])

reversed_text = "".join(characters)

print("Original string:", text)
print("Reversed string:", reversed_text)
```

## 8. Remove duplicates from a list

```python
values = [int(item) for item in input("Enter numbers separated by space: ").split()]

seen = set()
without_duplicates = []

for value in values:
    if value not in seen:
        seen.add(value)
        without_duplicates.append(value)

print("Original list:", values)
print("List after removing duplicates:", without_duplicates)
```

## 9. Merge two dictionaries

```python
first = {"a": 10, "b": 20}
second = {"c": 30, "d": 40}

combined = first.copy()
combined.update(second)

print("First dictionary:", first)
print("Second dictionary:", second)
print("Merged dictionary:", combined)
```

## 10. Fibonacci series using recursion

```python
def fib_term(index):
    if index < 2:
        return index
    return fib_term(index - 1) + fib_term(index - 2)

terms = int(input("Enter number of terms: "))
sequence = [fib_term(index) for index in range(terms)]

print("Fibonacci series:", *sequence)
```
