## Question 1 – Function calculate(a, b, operation)

```python
def calculate(a, b, operation):
    if operation == "/":
        return a / b if b != 0 else "Cannot divide by zero"

    operations = {
        "+": lambda: a + b,
        "-": lambda: a - b,
        "*": lambda: a * b
    }

    return operations[operation]() if operation in operations else "Invalid operation"


first = float(input("Enter first number: "))
second = float(input("Enter second number: "))
symbol = input("Enter operation (+, -, *, /): ")

print("Result:", calculate(first, second, symbol))
```

## Question 2 – Function sum_numbers(*args)

```python
from functools import reduce

def sum_numbers(*args):
    if not args:
        return 0
    return reduce(lambda total, value: total + value, args)

values = [int(value) for value in input("Enter numbers separated by space: ").split()]

print("Numbers:", values)
print("Sum:", sum_numbers(*values))
```

## Question 3 – Function employee(**args)

```python
def employee(**details):
    print("Employee Information:")
    for field in details:
        print(f"{field}: {details[field]}")


employee(
    name="Moulidhar",
    ID="EMP101",
    department="IT",
    salary=50000
)
```

## Question 4 – Function remove_duplicates(lst)

```python
def remove_duplicates(lst):
    return list(dict.fromkeys(lst))


numbers = [int(value) for value in input("Enter numbers separated by space: ").split()]

result = remove_duplicates(numbers)

print("Original list:", numbers)
print("List after removing duplicates:", result)
```

## Question 5 – Lambda Sort by Second Element

```python
numbers = [(1, 5), (2, 3), (4, 1)]

sorted_numbers = sorted(numbers, key=lambda pair: pair[-1])

print("Original list:", numbers)
print("Sorted list based on second element:", sorted_numbers)
```
