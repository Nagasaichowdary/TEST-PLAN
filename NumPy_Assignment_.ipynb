## Question 1 – Student Marks Array

```python
import numpy as np

marks = np.array([78, 65, 89, 56, 92])

print("Marks:", marks)
print("Dimensions:", marks.ndim)
print("Shape:", tuple(marks.shape))
print("Number of elements:", marks.size)
print("Type of data:", marks.dtype)
```

## Question 2 – Student Marks Access

```python
import numpy as np

scores = np.array([72, 85, 64, 90, 76])

print("Marks:", scores)
print("First student:", scores.take(0))
print("Third student:", scores.take(2))
print("Last student:", scores[-1])
print("First three:", scores[:3])
print("Second through fourth:", scores[1:4])
print("Last two:", scores[-2:])
```

## Question 4 – Internal and External Marks

```python
import numpy as np

internal = np.array([20, 18, 22, 19, 21])
external = np.array([65, 70, 68, 72, 75])

total = np.add(internal, external)

print("Internal:", internal)
print("External:", external)
print("Combined marks:", total)
```

## Question 5 – Pass Percentage Analysis

```python
import numpy as np

marks = np.array([45, 78, 56, 32, 91])

qualified = marks[marks >= 50]

print("Marks:", marks)
print("Pass condition:", marks >= 50)
print("Students who passed:", qualified)
print("Number passed:", qualified.size)
```

## Question 6 – Average Marks

```python
import numpy as np

marks = np.array([
    [78, 85, 90],
    [65, 72, 80],
    [88, 91, 84],
    [56, 62, 70],
    [95, 89, 92]
])

student_averages = marks.sum(axis=1) / marks.shape[1]

print("Marks:")
print(marks)
print("Average of each student:")
print(student_averages)
```

## Question 7 – Class Performance Statistics

```python
import numpy as np

scores = np.array([67, 82, 91, 74, 58])

summary = {
    "Total": scores.sum(),
    "Average": scores.mean(),
    "Highest": scores.max(),
    "Lowest": scores.min(),
    "Standard Deviation": scores.std()
}

print("Marks:", scores)
for label, value in summary.items():
    print(label + ":", value)
```

## Question 10 – Student Ranking

```python
import numpy as np

marks = np.array([245, 278, 219, 290, 256])

order = np.argsort(-marks)

print("Original marks:", marks)
print("Descending marks:", marks[order])
print("Ranking:")

for position, student_index in enumerate(order, 1):
    print(f"Rank {position} - Student {student_index + 1} - Marks: {marks[student_index]}")
```

## Question 11 – Duplicate Marks Analysis

```python
import numpy as np

marks = np.array([85, 92, 85, 76, 92])

unique_marks, counts = np.unique(marks, return_counts=True)

print("Original marks:", marks)
print("Marks appearing more than once:", unique_marks[counts > 1])
print("All unique marks:", unique_marks)
```

## Question 12 – Missing Marks

```python
import numpy as np

marks = np.array([78, 85, np.nan, 92, 67])

valid_marks = marks[~np.isnan(marks)]
average = valid_marks.mean()

print("Marks:", marks)
print("Available marks:", valid_marks)
print("Average of available marks:", average)
```

## Question 13 – Grade Classification

```python
import numpy as np

marks = np.array([95, 82, 74, 61, 45])

grades = np.full(marks.shape, "F", dtype="<U1")
grades[marks >= 60] = "D"
grades[marks >= 70] = "C"
grades[marks >= 80] = "B"
grades[marks >= 90] = "A"

print("Marks:", marks)
print("Grades:", grades)
```

## Question 14 – Random Marks Generation

```python
import numpy as np

rng = np.random.default_rng(10)
marks = rng.integers(0, 101, size=5)

print("Generated marks:", marks)
print("Total:", marks.sum())
print("Average:", marks.mean())
print("Highest:", marks.max())
print("Lowest:", marks.min())
print("Standard deviation:", marks.std())
```

## Question 15 – Student Performance Analysis

```python
import numpy as np

marks = np.array([
    [78, 85, 90],
    [65, 72, 80],
    [88, 91, 84],
    [56, 62, 70],
    [95, 89, 92]
])

totals = marks.sum(axis=1)
averages = marks.mean(axis=1)
highest_scores = marks.max(axis=1)
lowest_scores = marks.min(axis=1)
class_average = marks.mean()

print("Student Marks:")
print(marks)
print("Total:", totals)
print("Average:", averages)
print("Highest:", highest_scores)
print("Lowest:", lowest_scores)
print("Class Average:", class_average)

above = np.flatnonzero(averages > class_average)
print("Students above class average:")

for student in above:
    print(f"Student {student + 1} - Average: {averages[student]}")
```
